```python
print("=" * 60)
print("         Day 3: FUNCTIONS MASTERCLASS")
print("=" * 60)
print()

print("Part1: Basic Functions\n")

def greet():
    """Asimple function with no parametes"""
    print("Hello, AI learner!")
    print("Ready to master functions? Let's go!")

greet()
print()

def greet_person(name):
  print(f"Hello, {name}! Welcome to day 3.")

greet_person("Future AI Engineer")
greet_person("Wangari")
print()

def calculate_study_time(hours_per_day, days):
  total = hours_per_day * days
  print(f"{hours_per_day} hours/day * {days} days = {total} hours")

calculate_study_time(2, 330)
calculate_study_time(3, 330)
calculate_study_time(6, 330)
print()

print("Part 2: Functions with Return Values\n")

def multiply (a, b):
  result = a * b
  return result

answer = multiply(7, 8)
print(f"7 * 8 = {answer}")

print(f"5 * 12 = {multiply(5, 12)}")
print()

def get_skill_level(hours):
    """Determine skill level based on hours invested"""
    if hours < 50:
        return "🌱 Beginner"
    elif hours < 150:
        return "📈 Novice"
    elif hours < 300:
        return "⚡ Intermediate"
    elif hours < 600:
        return "🔥 Advanced"
    else:
        return "🏆 Expert"

my_hours = 6
print(f"at {my_hours} hours: {get_skill_level(my_hours)}")
print(f"At 100 hours: {get_skill_level(100)}")
print(f"At 500 hours: {get_skill_level(500)}")
print()


print("Part 3: Default Parameters\n")

def log_progress(day, hours=2, feeling="motivated"):
  print(f"Day {day}: Studied {hours} hours, Feeling {feeling}")

log_progress(1)
log_progress(2)

log_progress(3, 2, "focused")
log_progress(4, 6, "unstoppable")
print()

print("PART 4: Multiple Return Values\n")

def analyze_progress(days, total_hours):
  avg_hours = total_hours / days
  projected_total = avg_hours * 330
  completion_percentage = (days / 330) * 100

  return avg_hours, projected_total, completion_percentage

avg, projected, percent = analyze_progress(3, 6)
print(f"📊 Your Progress Analysis:")
print(f"  Average hours/day: {avg:.1f}")
print(f"  Projected 11-month total: {projected:.0f} hours")
print(f"  Journey completion: {percent:.1f}%")
print()


print("PART 5: Your AI Learning Toolkit\n")

def days_remaining(current_day):
    """Calculate days left in journey"""
    return 330 - current_day

def hours_to_milestone(current_hours, milestone):
    """How many hours until next milestone"""
    return milestone - current_hours

def predict_date(current_day, hours_per_day, target_hours):
    """Predict when you'll reach target hours"""
    hours_so_far = current_day * hours_per_day
    hours_needed = target_hours - hours_so_far
    days_needed = hours_needed / hours_per_day # Corrected typo here
    return int(days_needed)

def motivation_level(days_completed):
    """Get motivational message based on progress"""
    if days_completed < 7:
        return "🔥 Strong start! Build the habit!"
    elif days_completed < 30:
        return "💪 First month momentum! Keep going!"
    elif days_completed < 100:
        return "⚡ You're in the zone! Don't stop!"
    else:
        return "🏆 Unstoppable! You're nearly there!"

current_day = 3
total_hours = 6
avg_daily_hours = 2

print(f"🗓️  Days remaining: {days_remaining(current_day)}")
print(f"⏱️  Hours to 100-hour milestone: {hours_to_milestone(total_hours, 100)}")
print(f"📅 Days until 500 hours: {predict_date(current_day, avg_daily_hours, 500)}")
print(f"✨ {motivation_level(current_day)}")
print()

print("Part 6: Complete progress calculator\n")

def full_progress_report(day, hours_today, total_hours, projects_done):

    avg_hours_per_day = total_hours / day
    days_remaining_calc = 330 - day # Renamed to avoid conflict with days_remaining function
    projected_hours = avg_hours_per_day * 330
    completion_percentage = (day / 330) * 100
    skill = get_skill_level(total_hours)

    report = f"""
    Day {day} Progress Report:
    current Day:{day} of 330
    Days remaining: {days_remaining_calc}
    completion percentage: {completion_percentage:.1f}%

    Today's hours: {hours_today}
    Total hours: {total_hours}
    Average hours per day: {avg_hours_per_day:.1f}
    Projected Total : {projected_hours:.0f}

    Completed Projects: {projects_done}
    Pace: {projects_done/day: .2f} per day

    Skill level: {skill}

    {motivation_level(day)}
    """
    return report

my_report = full_progress_report(
    day=3,
    hours_today=2,
    total_hours=6,
    projects_done=2
)
print(my_report)

with open("progress_report.txt", "w") as file:
    file.write(my_report)
    print("Report Saved to 'progress_report.txt'\n")
```
