```python
training_data = [
    {"experience_years": 0, "salary_kes": 150000},
    {"experience_years": 1, "salary_kes": 250000},
    {"experience_years": 2, "salary_kes": 380000},
    {"experience_years": 3, "salary_kes": 520000},
    {"experience_years": 4, "salary_kes": 650000},
    {"experience_years": 5, "salary_kes": 786000}
]
print("TRAINING DATA:")
print(f"{'Experience':15} {'Salary(KES)':15}")
print("-" * 30)
for row in training_data:
 print(f"{str(row['experience_years']) + ' years':15} {row['salary_kes']:,}")
total_increase = training_data[-1]["salary_kes"] - training_data[0]["salary_kes"]
total_years = training_data[-1]["experience_years"] - training_data[0]["experience_years"]
salary_increase_per_year = total_increase / total_years
base_salary = training_data[0]["salary_kes"]

print(f"\n PATTERN FOUND:")
print(f" Base Salary(0 experience): KES{base_salary:,}")
print(f" Salary increase per year: KES{salary_increase_per_year:,}")

def salary_model(years_experience):
  """Our trained salary prediction model"""
  predicted_salary = base_salary + (salary_increase_per_year * years_experience)
  return round(predicted_salary)

print(f"\n PREDICTION:")
test_cases = [1.5, 2.5, 3, 6, 10]
for years in test_cases:
  prediction = salary_model(years)
  print(f" {years} years experience - KES{prediction:,}/month")
print("\n This is EXACTLY what scikit-learn does")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

print("✅ All libraries imported successfully!")
print("✅ You are ready to build your first ML model.")
print("💙 Welcome to Machine Learning, Wangari.")
```

```python
X = np.array([0, 0.5, 1, 1.5, 2, 2.5, 3, 4, 5]).reshape(-1, 1)
y = np.array([150000, 200000, 250000, 315000, 380000, 450000, 520000, 650000, 7860000])

print("Training Data Ready:")
print(f" X shape: {X.shape} - {len(X)} experience data points")
print(f" y shape: {y.shape} - {len(y)} salary data points")
print(f"\n  First 3 examples:")
for i in range(3):
  print(f"   {X[i][0]} years experience → KES {y[i]:,}/month")
```

```python
model = LinearRegression()
model.fit(X, y)
print("🎉 Model trained successfully!")
print(f"\n🧠 What the model learned:")
print(f"   Base salary (0 experience): KES {model.intercept_:,.0f}")
print(f"   Salary increase per year:   KES {model.coef_[0]:,.0f}")
print(f"\n   Formula: Salary = {model.intercept_:,.0f} + ({model.coef_[0]:,.0f} × years)")
print(f"\n   Compare to our manual calculation in Hour 5!")
print(f"   scikit-learn finds the mathematically OPTIMAL line.")
```

```python
future_years = np.array([1.5, 2, 3, 5, 7, 10]). reshape(-1, 1)
prediction = model.predict(future_years)

print(" WANGARI'S SALARY PREDICTIONS:")
print("="*50)
for years, salary in zip(future_years, prediction):
  year_label = f"{years[0]} years"
  print(f"  {year_label:12} → KES {salary:>12,.0f}/month")
print("="*45)

# How accurate is our model?
y_predicted = model.predict(X)
mae = mean_absolute_error(y, y_predicted)
r2 = r2_score(y, y_predicted)

print(f"\n📊 MODEL ACCURACY:")
print(f"   Mean Absolute Error: KES {mae:,.0f}")
print(f"   R² Score: {r2:.4f} (1.0 = perfect, 0.0 = useless)")
if r2 > 0.95:
    print(f"   ✅ Excellent model! Very accurate predictions.")
elif r2 > 0.80:
    print(f"   ✅ Good model. Reliable predictions.")
else:
    print(f"   ⚠️  Model needs more or better training data.")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='#f0b429', s=100, zorder=5, label='Real Salaries')
X_line = np.linspace(0, 10, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='#2dd4bf', linewidth=2.5, label='Model Predictions')

Wangari_years = [0, 1, 3, 5]
Wangari_salary = model.predict(np.array(Wangari_years).reshape(-1, 1))
for yr, sal in zip(Wangari_years, Wangari_salary):
  plt.annotate(f'KES {sal:,.0f}', (yr, sal),
                textcoords="offset points", xytext=(8,8),
                fontsize=9, color='white')

plt.xlabel('Years of ML Experience', fontsize=12)
plt.ylabel('Monthly Salary (KES)', fontsize=12)
plt.title("Wangari's ML Career Salary Trajectory 🚀", fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'KES {x:,.0f}'))
plt.tight_layout()
plt.show()

print("\n🎉 You just built, trained, and visualised your first ML model!")
print("💙 That chart shows YOUR future. You built it. You own it.")
```
