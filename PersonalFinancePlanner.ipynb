```python
def calculate_expenses(rent, food, internet, transport):
    """Adds up all monthly expenses"""
    total = rent + food + internet + transport
    return total


def calculate_savings(income, expenses):
    """Calculates monthly savings"""
    savings = income - expenses
    return savings


def months_to_goal(monthly_savings, goal_amount):
    """Calculates how many months to reach a financial goal"""
    if monthly_savings <= 0:
        return "Cannot reach goal - expenses exceed income"
    months = goal_amount / monthly_savings
    return round(months, 1)


def financial_report(name, income, rent, food, internet, transport):
    """Complete Financial Health Report"""

    expenses = calculate_expenses(rent, food, internet, transport)
    savings = calculate_savings(income, expenses)

    months_visa = months_to_goal(savings, 500000)
    months_emergency = months_to_goal(savings, 300000)

    print(f"\n{'='*45}")
    print(f"Financial Report - {name}")
    print(f"{'='*45}")
    print(f"Monthly Income:   KES {income:,}")
    print(f"Monthly Expenses: KES {expenses:,}")
    print(f"Monthly Savings:  KES {savings:,}")
    print(f"Savings Rate:     {(savings/income*100):.1f}%")
    print(f"Emergency Fund:   {months_emergency} months away")
    print(f"UK Visa Fund:     {months_visa} months away")

    if savings > 0:
        print("Status: ✅ SURPLUS — You are building wealth!")
    else:
        print("Status: ⚠️ DEFICIT — Income needs to increase")

    print(f"{'='*45}")


print("🗓️  WANGARI'S FINANCIAL JOURNEY")

financial_report(
    "Today — AI Trainer Income",
    income=120000, rent=30000, food=25000, internet=3000, transport=5000
)

financial_report(
    "Year 1 — Junior ML Engineer",
    income=250000, rent=40000, food=30000, internet=3000, transport=8000
)

financial_report(
    "Year 3 — Mid Level ML Engineer",
    income=786000, rent=80000, food=40000, internet=5000, transport=15000
)
```
