```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n = 200

data = {
    "Name_ID": [f"Worker_{i}" for i in range(1, n+1)],
    "Role": np.random.choice(
        ["ML Engineer", "Data Analyst", "Software Dev", "Data Scientist",  "AI Trainer"], n),
    "Experience_Years": np.random.randint(0, 12, n),
    "City": np.random.choice(
        ["Nairobi", "Mombasa", "Kisumu", "Nakuru"], n, p=[0.6, 0.2, 0.1, 0.1]),
    "Remote": np.random.choice([True, False], n, p=[0.4, 0.6]),
    "Has_Degree": np.random.choice([True, False], n, p=[0.7, 0.3])
}

base_salaries = {
    "ML Engineer": 120000,
    "Data Scientist": 110000,
    "Software Dev": 100000,
    "Data Analyst": 90000,
    "AI Trainer": 80000
}

salaries = []
for i in range(n):
  base = base_salaries[data["Role"][i]]
  exp_bonus = data["Experience_Years"][i] * 15000
  remote_bonus = 40000 if data["Remote"][i]else 0
  noise = np.random.randint(-10000, 10000)
  salaries.append(base + exp_bonus + remote_bonus + noise)
data["Monthly_Salary_KES"] = salaries

df = pd.DataFrame(data)
print("=== DATASET INFO ===")
print(df.info())
print()

print("=== AVERAGE SALARY BY ROLE ===")
avg_by_role = (
    df.groupby("Role")["Monthly_Salary_KES"]
      .mean()
      .sort_values(ascending=False)
    )
print(avg_by_role.apply(lambda x: f"KES{x:,.0F}"))
print()

print("=== REMOTE Vs OFFICE SALARY ===")
remote_compare = df.groupby("Remote")["Monthly_Salary_KES"].mean()

for remote, salary in remote_compare.items():
  label = "Remote" if remote else "Office"
  print(f"{label}: KES {salary:,.0f}")
print()

print("=== ML ENGINEERS ONLY ===")
ml_engineers = df[df["Role"] == "ML Engineer"]

print(f"Count: {len(ml_engineers)}")
print(f"Average: KES {ml_engineers['Monthly_Salary_KES'].mean():,.0f}")
print(f"Highest: KES {ml_engineers['Monthly_Salary_KES'].max():,.0f}")
print(f"Lowest: KES {ml_engineers['Monthly_Salary_KES'].min():,.0f}")
```
