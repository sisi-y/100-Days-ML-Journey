```python
def calculate_monthly_income(hours_per_day, rate_per_hour, working_days=22):
  """
  Calculates the monthly income.
  hours_per_day: Number of hours you work daily
  rate_per_hour: Your hourly rate in KES
  working_days: defaults to 22 in not specified
  """
  daily_income = hours_per_day * rate_per_hour
  monthly_income = daily_income * working_days
  return monthly_income
entry_level = calculate_monthly_income(6, 600)
mid_level = (calculate_monthly_income(6, 1200))
senior = (calculate_monthly_income(6, 3000, 20))

print(f"Entry Level AI Trainer:KES{entry_level:,}/month")
print(f"Mid Level AI Trainer: KES{mid_level:,}/month")
print(f"senior ML Engineer: KES {senior:,}/month")
```

```python
def calculate_tax(income):
  """Calculates approximate Kenya PAYE tax"""
  if income <=24000:
    return income * 0.10
  elif income <= 32333:
    return income * 0.25
  else:
    return income * 0.30
def calculate_net_salary(gross_salary):
  """Calculate take-home pay after tax"""
  tax = calculate_tax(gross_salary)
  net = gross_salary - tax
  return net, tax
def salary_report(name, gross):
  """Full salary breakdown report"""
  net, tax = calculate_net_salary(gross)
  print(f"\n💰 Salary Report for {name}")
  print(f"   Gross Salary: KES {gross:,}")
  print(f"   Tax Deducted: KES {tax:,.0f}")
  print(f"   Net Take-Home: KES {net:,.0f}")
  if net > 500000:
        print(f"   Status: 🏆 HIGH EARNER — Living the dream!")
  elif net > 150000:
        print(f"   Status: ✅ COMFORTABLE — Above your monthly target")
  else:
        print(f"   Status: 📈 BUILDING — Keep growing your skills")

# Generate reports for different career stages
salary_report("Wangari — Year 1", 200000)
salary_report("Wangari — Year 3", 786000)
salary_report("Wangari — Year 5", 1200000)

```
