```python
study_hours = [5, 6, 0, 4, 6, 5, 0, 5, 4, 6, 6, 4, 5, 5]
total = 0
best_day = 0
rest_days = 0
streak = 0
best_streak = 0
current_streak = 0

for i, hours in enumerate(study_hours):
  total += hours
  if hours > best_day:
    best_day = hours
  if hours == 0:
    rest_days += 1
    current_streak = 0
  else:
    current_streak > best_streak
    best_streak = current_streak

average = total / len(study_hours)

print("📊 ====== YOUR STUDY ANALYSIS ====== 📊")
print(f"📅 Days tracked:     {len(study_hours)}")
print(f"⏰ Total hours:      {total}")
print(f"📈 Daily average:    {average:.1f} hours")
print(f"🏆 Best day:         {best_day} hours")
print(f"🔥 Best streak:      {best_streak} days in a row")
print(f"😴 Rest days:        {rest_days}")
print("====================================")

if average >= 5:
    print("💪 You are on track for your UK goal!")
elif average >= 3:
    print("📈 Good progress! Push for 5 hours daily.")
else:
    print("⚡ More consistency needed. You've got this!")
```
