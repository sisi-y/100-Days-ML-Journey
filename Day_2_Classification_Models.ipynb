```python
students = ["Alice", "Bob", "Charlie", "Diana", "Eve",
            "Frank", "Grace", "Henry", "Iris", "Jack"]
study_hours = [8, 4, 6, 2, 7, 3, 9, 5, 1, 6]      # Hours studied
previous_score = [75, 55, 68, 45, 80, 50, 85, 60, 40, 70]  # Previous exam %
attendance = [90, 70, 85, 60, 95, 65, 98, 75, 55, 88]
results = [1, 1, 1, 0, 1, 0, 1, 1, 0, 1]

print("OUR STUDENT DATA:")
print("=" * 80)
print(f"{'Student':<10} {'Study Hrs':<12}{'Previous Score':<12} {'Attendance':<12}")
print("=" * 80)

for i in range(len(students)):
    result_text = "✅ PASSED" if results[i] == 1 else "❌ FAILED"
    print(f"{students[i]:<10} {study_hours[i]:<12} {previous_score[i]:<12} {attendance[i]:<12} {result_text:<10}")

print("=" * 80)
print()
print(f"Total students: {len(students)}")
print(f"Passed: {sum(results)} students")
print(f"Failed: {len(results) - sum(results)} students")
```

```python
X = np.array([study_hours, previous_score, attendance])
X = X.T
print("INPUT DATA (X):")
print("=" * 60)
print("Each row = one student")
print("Each column = one piece of information")
print()
print(X)
print()
print(f"Shape: {X.shape}")
print(f"That means: {X.shape[0]} students, {X.shape[1]} features")
print()

y = np.array(results)

print("OUTPUT DATA (y):")
print("=" * 60)
print("1 = Passed, 0 = Failed")
print()
print(y)
print()
print(f"Shape: {y.shape}")
print(f"That means: {y.shape[0]} results (one per student)")
```
