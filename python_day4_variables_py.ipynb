```python
skills_to_learn = [
    "Python",
    "Math for ML",
    "Machine Learning",
    "Deep Learning",
    "Neural Networks",
    "NLP",
    "Computer Vision",
    "Mlops",
    "Portfolio Building",
    "Job Application"
]
print("MY AI\ML LEARNING PATH")
print("=" * 50)
for i in range(len(skills_to_learn)):
  print(f"{i + 1}. {skills_to_learn[i]}")

skills_started = ["Python"]
skills_completed = []

total_skills = len(skills_to_learn)
started = len(skills_started)
completed = len(skills_completed)

print("\n" + "=" * 50)
print("PROGRESS TRACKER")
print(f"Total skils to learn: {total_skills}")
print(f"Skills started: {started}")
print(f"Skills completed: {completed}")
print(f"Progress: {(completed / total_skills) * 100:.1f}%")

current_skill_index = len(skills_completed)
if current_skill_index < total_skills:
    next_skill = skills_to_learn[current_skill_index]
    print(f"\nNext skill to focus on: {next_skill}")

# Motivation
remaining = total_skills - completed
print(f"\nOnly {remaining} skills to go until job-ready!")
print("You got this! 💪")
```
